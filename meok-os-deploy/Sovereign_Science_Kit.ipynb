## Governance lookup + LIVE trust score (os.meok.ai, serverless, free)

```
import requests
B='https://os.meok.ai'
print('governs a bank →', [f['name'] for f in requests.get(B+'/api/govern',params={'q':'I run a bank'}).json()['frameworks'][:4]])
print('trust(Aria#owl) →', requests.get(B+'/api/trust/score/Aria%23owl').json()['tier'])
```

## The MOAT: sign a research claim → offline-verifiable provenance record

```
claim='Metformin reduces HbA1c in T2DM (illustrative)'
src='doi:10.1001/example'
prov=requests.get(B+'/api/provenance',params={'claim':claim,'source':src,'kind':'trial'}).json()
sig=prov['signature']
print('signed fingerprint:', sig['fingerprint'])
# verify it — anyone, offline
v=requests.post(B+'/api/verify',json={'message':sig['canonical'],'signature':sig['signature'],'publicKey':sig['publicKey']}).json()
print('verifies:', v['valid'], '·', v.get('message'))
```
