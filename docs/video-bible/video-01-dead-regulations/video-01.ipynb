## 1. Narration track

```
# Video 1: THE PROBLEM — DEAD REGULATIONS — narration (gTTS draft; re-record on phone for final)
!pip install -q gTTS

from gtts import gTTS
import IPython.display as ipd

script = """
The EU AI Act just passed.
It's already dead.
A regulator writes Article 52.
By the time it's published, 3 new AI models have launched.
By the time a company reads it, 12 new attack vectors exist.
By the time a CISO trains their staff, the training is obsolete.
The old world:
Static PDFs. Frozen certifications. £100,000 compliance courses
that expire the moment you finish them.
We don't need more regulations.
We need LIVING governance.
SOVOS.
The first operating system that governs AI...
while AI is still changing.
"""

tts = gTTS(text=script, lang="en", slow=False)
tts.save("video01_narration.mp3")
ipd.Audio("video01_narration.mp3")
```

## 2. Keyframe images

```
# Video 1: THE PROBLEM — DEAD REGULATIONS — keyframe images (Stable Diffusion, free Colab T4)
!pip install -q diffusers transformers accelerate

from diffusers import StableDiffusionPipeline
import torch

pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16
).to("cuda")

prompts = [
    "EU Parliament building, golden hour, cinematic, 4K",
    "Calendar pages dissolving into dust particles, dark background",
    "Exhausted executive at desk buried in papers, photorealistic",
    "Graveyard of paper certificates crumbling, moody lighting",
    "Glowing futuristic OS interface, dark background, gold and blue accents"
]

for i, prompt in enumerate(prompts, 1):
    image = pipe(prompt).images[0]
    image.save(f"video01_image{i}.png")
    print(f"saved video01_image{i}.png — {prompt}")
```

## 3. Assemble

Drop the MP3 + PNGs into CapCut/iMovie, or `ffmpeg -framerate 1/8 -i video%02d_image%d.png -i narration.mp3 -c:v libx264 -pix_fmt yuv420p out.mp4`.
