## 1. Narration track

```
# Video 4: THE SANDWICH — FLUID INFERENCE — narration (gTTS draft; re-record on phone for final)
!pip install -q gTTS

from gtts import gTTS
import IPython.display as ipd

script = """
The AI Sandwich.
Top: The Cloud.
Kimi K3. Claude Opus. DeepSeek.
Big. Expensive. Sometimes needed.
Middle: The Router.
SOVOS decides: cloud or local? Fast or deep?
Every decision GSPC-scored.
The Cheese: Honey.
Knowledge distilled from every interaction.
The system learns while you sleep.
The Patty: Local.
Gemma 4 on your Mac. SOV3 on your farm.
Private. Fast. Sovereign.
Bottom: The Edge.
A 1-billion-parameter model on your TV.
Always on. Always local. Always free.
But this sandwich is FLUID.
It grows when you need power.
It shrinks when you need privacy.
It adapts like an insect brain switching gears.
"""

tts = gTTS(text=script, lang="en", slow=False)
tts.save("video04_narration.mp3")
ipd.Audio("video04_narration.mp3")
```

## 2. Keyframe images

```
# Video 4: THE SANDWICH — FLUID INFERENCE — keyframe images (Stable Diffusion, free Colab T4)
!pip install -q diffusers transformers accelerate

from diffusers import StableDiffusionPipeline
import torch

pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16
).to("cuda")

prompts = [
    "Gourmet hamburger transforming into layered tech architecture, surreal",
    "Glowing cloud servers, futuristic, dark background",
    "Golden honey flowing between layers, luminous",
    "Apple Silicon chip with dragon emblem, photorealistic",
    "Tiny glowing orb on TV screen, minimalist",
    "Liquid metal transforming between states, cinematic"
]

for i, prompt in enumerate(prompts, 1):
    image = pipe(prompt).images[0]
    image.save(f"video04_image{i}.png")
    print(f"saved video04_image{i}.png — {prompt}")
```
