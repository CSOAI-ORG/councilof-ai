## 2. Keyframe images

```
# Video 6: THE SWARM — OWEM INSECTS — keyframe images (Stable Diffusion, free Colab T4)
!pip install -q diffusers transformers accelerate

from diffusers import StableDiffusionPipeline
import torch

pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16
).to("cuda")

prompts = [
    "Macro photography of ant, then zoom out to millions, cinematic",
    "Four biomechanical insects: scout (yellow), worker (green), guardian (blue), queen (gold)",
    "Insect dissolving into golden particles, beautiful, ethereal",
    "Golden queen insect breathing life into spawn, majestic"
]

for i, prompt in enumerate(prompts, 1):
    image = pipe(prompt).images[0]
    image.save(f"video06_image{i}.png")
    print(f"saved video06_image{i}.png — {prompt}")
```
