## 1. Narration track

```
# Video 3: THE 4 AXES — GSPC — narration (gTTS draft; re-record on phone for final)
!pip install -q gTTS

from gtts import gTTS
import IPython.display as ipd

script = """
Every AI decision is a direction.
Most companies navigate blind.
G — Governance.
Is this lawful? Does it comply with EU AI Act, NIST, ISO?
S — Security.
Is this safe? Will it leak? Can it be hacked?
P — Privacy.
Does this respect the human? Is data sovereign?
C — Commerce.
Is this profitable? Is it efficient?
Every agent in SOVOS carries a GSPC score.
Every swarm votes with these scores.
Every human sees them on their TV.
You don't trust AI because someone told you to.
You trust it because you can SEE it.
"""

tts = gTTS(text=script, lang="en", slow=False)
tts.save("video03_narration.mp3")
ipd.Audio("video03_narration.mp3")
```

## 2. Keyframe images

```
# Video 3: THE 4 AXES — GSPC — keyframe images (Stable Diffusion, free Colab T4)
!pip install -q diffusers transformers accelerate

from diffusers import StableDiffusionPipeline
import torch

pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16
).to("cuda")

prompts = [
    "Ancient compass glowing with 4 points, futuristic, gold",
    "Four geometric pillars: Governance, Security, Privacy, Commerce",
    "Rotating tetrahedron in 4D space, wireframe, glowing edges",
    "Smart TV screen with glowing HUD overlay, living room background"
]

for i, prompt in enumerate(prompts, 1):
    image = pipe(prompt).images[0]
    image.save(f"video03_image{i}.png")
    print(f"saved video03_image{i}.png — {prompt}")
```
