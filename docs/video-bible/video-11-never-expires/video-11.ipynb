## 1. Narration track

```
# Video 11: THE LIVING CERT — NEVER EXPIRES — narration (gTTS draft; re-record on phone for final)
!pip install -q gTTS

from gtts import gTTS
import IPython.display as ipd

script = """
This certificate cost £100,000.
It took 6 months to earn.
It expired yesterday.
Static certifications are dead.
The world changes faster than paper can record.
This is a Living Certificate.
EU AI Act changes?
A dragon appears over your castle.
Defeat it in a 5-minute simulation.
Your certificate updates.
Instantly.
Blockchain-anchored.
Court-admissible.
Level 1: Foundation.
Level 5: Practitioner.
Level 10: Master.
Level 20: Sovereign.
Regulators don't ask "When did you train?"
They ask "Are you current?"
With a Living Cert, the answer is always YES.
"""

tts = gTTS(text=script, lang="en", slow=False)
tts.save("video11_narration.mp3")
ipd.Audio("video11_narration.mp3")
```

## 2. Keyframe images

```
# Video 11: THE LIVING CERT — NEVER EXPIRES — keyframe images (Stable Diffusion, free Colab T4)
!pip install -q diffusers transformers accelerate

from diffusers import StableDiffusionPipeline
import torch

pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16
).to("cuda")

prompts = [
    "Expensive paper certificate crumbling, dust particles",
    "Red EXPIRED stamp, dramatic",
    "Glowing golden certificate on medieval castle wall",
    "Player fighting red dragon, UE5 style, epic",
    "Certificate evolving through levels, RPG style",
    "Regulator with tablet, green checkmark, satisfied"
]

for i, prompt in enumerate(prompts, 1):
    image = pipe(prompt).images[0]
    image.save(f"video11_image{i}.png")
    print(f"saved video11_image{i}.png — {prompt}")
```
