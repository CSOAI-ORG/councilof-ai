## 1. Narration track

```
# Video 8: THE PORTAL — TV AS SOVEREIGN NODE — narration (gTTS draft; re-record on phone for final)
!pip install -q gTTS

from gtts import gTTS
import IPython.display as ipd

script = """
You already own the most powerful
AI governance device on Earth.
Your TV.
It has a microphone.
A camera. WiFi. A GPU.
And now... a conscience.
"Hey SOV, what's our home score?"
"Hey SOV, did the AI Act change?"
"Hey SOV, is this email a scam?"
Your kids learn AI safety by playing.
Your parents stay protected without touching a laptop.
You govern your home without reading a 400-page PDF.
And in the Security Operations Center?
The same TV.
The same interface.
The same mind.
"""

tts = gTTS(text=script, lang="en", slow=False)
tts.save("video08_narration.mp3")
ipd.Audio("video08_narration.mp3")
```

## 2. Keyframe images

```
# Video 8: THE PORTAL — TV AS SOVEREIGN NODE — keyframe images (Stable Diffusion, free Colab T4)
!pip install -q diffusers transformers accelerate

from diffusers import StableDiffusionPipeline
import torch

pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16
).to("cuda")

prompts = [
    "Family entering modern living room, TV off, warm lighting",
    "Futuristic TV interface glowing, dark room, blue and gold",
    "Voice waveforms on screen, minimalist, glowing",
    "Kids playing fantasy game on big TV, happy, colorful",
    "Security operations center with massive video wall, cinematic"
]

for i, prompt in enumerate(prompts, 1):
    image = pipe(prompt).images[0]
    image.save(f"video08_image{i}.png")
    print(f"saved video08_image{i}.png — {prompt}")
```
