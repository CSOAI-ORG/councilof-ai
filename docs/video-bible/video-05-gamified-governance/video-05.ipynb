## 2. Keyframe images

```
# Video 5: THE ARENA — GAMIFIED GOVERNANCE — keyframe images (Stable Diffusion, free Colab T4)
!pip install -q diffusers transformers accelerate

from diffusers import StableDiffusionPipeline
import torch

pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16
).to("cuda")

prompts = [
    "Boring corporate training room, people asleep, desaturated",
    "Explosion of color into fantasy city, UE5 style, vibrant",
    "Medieval castle under attack by red dragons, cinematic",
    "Gothic tower glowing blue, magical, night scene",
    "Happy village with villagers, fantasy art style",
    "Blacksmith forge with magical sparks, fantasy",
    "Family on couch playing TV game, warm lighting",
    "Glowing golden badge on stone wall, magical"
]

for i, prompt in enumerate(prompts, 1):
    image = pipe(prompt).images[0]
    image.save(f"video05_image{i}.png")
    print(f"saved video05_image{i}.png — {prompt}")
```
