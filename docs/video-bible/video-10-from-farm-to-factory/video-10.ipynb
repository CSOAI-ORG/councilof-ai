## 2. Keyframe images

```
# Video 10: CASA SECTORS — FROM FARM TO FACTORY — keyframe images (Stable Diffusion, free Colab T4)
!pip install -q diffusers transformers accelerate

from diffusers import StableDiffusionPipeline
import torch

pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16
).to("cuda")

prompts = [
    "Four minimalist icons: crane, wheat, shield, brain, geometric",
    "Construction site with futuristic AI crane, photorealistic",
    "High-tech greenhouse with drones, lush, vibrant",
    "Security operations center, dark, screens glowing",
    "Glowing AI brain, neural network, cinematic",
    "Four sectors merging into single golden node, abstract"
]

for i, prompt in enumerate(prompts, 1):
    image = pipe(prompt).images[0]
    image.save(f"video10_image{i}.png")
    print(f"saved video10_image{i}.png — {prompt}")
```

## 3. Assemble

Drop the MP3 + PNGs into CapCut/iMovie, or `ffmpeg -framerate 1/8 -i video%02d_image%d.png -i narration.mp3 -c:v libx264 -pix_fmt yuv420p out.mp4`.
