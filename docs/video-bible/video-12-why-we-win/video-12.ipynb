## 2. Keyframe images

```
# Video 12: THE FLYWHEEL — WHY WE WIN — keyframe images (Stable Diffusion, free Colab T4)
!pip install -q diffusers transformers accelerate

from diffusers import StableDiffusionPipeline
import torch

pipe = StableDiffusionPipeline.from_pretrained(
    "runwayml/stable-diffusion-v1-5", torch_dtype=torch.float16
).to("cuda")

prompts = [
    "Single person at computer, cozy, focused",
    "Human actions transforming into golden data particles",
    "Crowd of avatars joining swarm, vibrant, growing",
    "Enterprise executive looking at holographic data, impressed",
    "Golden flywheel spinning, mechanical and organic",
    "Competitors falling behind, abstract, motion blur",
    "Glowing SOVOS logo, majestic, final frame"
]

for i, prompt in enumerate(prompts, 1):
    image = pipe(prompt).images[0]
    image.save(f"video12_image{i}.png")
    print(f"saved video12_image{i}.png — {prompt}")
```
